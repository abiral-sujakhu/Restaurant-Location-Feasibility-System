# presence-vs-background classification

```python
import numpy as np
import pandas as pd
from scipy.spatial import cKDTree
from pyproj import Transformer

# ---- paths (edit to match your repo) ----
POI_PATH  = "../data/processed/pois_unique.csv"
REST_PATH = "../data/raw_data/restaurants_raw_multiarea.csv"

# ---- constants ----
RADIUS = 500.0          # metres: POI + competitor search radius
AREA_RADIUS = 1500.0    # metres: collection circle per area
EPSG_UTM = "EPSG:32645" # UTM 45N — real metres for Kathmandu

_to_utm = Transformer.from_crs("EPSG:4326", EPSG_UTM, always_xy=True)
_to_ll  = Transformer.from_crs(EPSG_UTM, "EPSG:4326", always_xy=True)
def to_xy(lat, lon):
    x, y = _to_utm.transform(np.asarray(lon, float), np.asarray(lat, float))
    return np.asarray(x), np.asarray(y)
def to_ll(x, y):
    lon, lat = _to_ll.transform(x, y)
    return float(lat), float(lon)

# 9 area centres
CENTERS = {
    "Baneshwor": (27.69396, 85.33738), "New Road": (27.70200, 85.30743),
    "Koteshwor": (27.68333, 85.35000), "Bhaktapur durbar square": (27.67203, 85.42811),
    "Patan durbar square": (27.67340, 85.32500), "Boudha stupa": (27.72139, 85.36194),
    "Pulchowk": (27.6787, 85.3175), "Durbar Marg": (27.71261, 85.31797),
    "Kirtipur": (27.67806, 85.27694),
}

# ---- build the feature engine once ----
pois = (pd.read_csv(POI_PATH, encoding="utf-8-sig")
          .dropna(subset=["latitude", "longitude"])
          .drop_duplicates("place_id"))
POI_TYPES = sorted(pois.poi_type.unique())
poi_trees = {}
for t in POI_TYPES:
    sub = pois[pois.poi_type == t]
    x, y = to_xy(sub.latitude.values, sub.longitude.values)
    poi_trees[t] = cKDTree(np.c_[x, y])

rest = pd.read_csv(REST_PATH).dropna(subset=["latitude", "longitude"])
R_RATING  = rest.restaurant_rating.values
R_REVIEWS = rest.user_rating_count.values
rx, ry = to_xy(rest.latitude.values, rest.longitude.values)
R_XY = np.c_[rx, ry]
rest_tree = cKDTree(R_XY)

def compute_features(lat, lon, exclude_self=False):
    """Neighbourhood features for ANY coordinate.
    exclude_self=True only when the point is itself a real restaurant."""
    x, y = to_xy(lat, lon)
    pt = np.array([float(x), float(y)])
    f = {}
    for t in POI_TYPES:
        f[f"{t}_count_500m"] = len(poi_trees[t].query_ball_point(pt, RADIUS))
    idx = rest_tree.query_ball_point(pt, RADIUS)
    if exclude_self:
        d = np.hypot(R_XY[idx, 0] - pt[0], R_XY[idx, 1] - pt[1])
        idx = [i for i, dd in zip(idx, d) if dd > 1e-6]
    rat, rev = R_RATING[idx], R_REVIEWS[idx]
    dd, _ = rest_tree.query(pt, k=2 if exclude_self else 1)
    near = max(dd) if exclude_self else float(dd)
    f["competitor_count_500m"] = len(idx)
    f["avg_restaurant_rating_500m"] = np.nanmean(rat) if len(rat) else np.nan
    f["avg_review_ratings_500m"]    = np.nanmean(rev) if len(rev) else np.nan
    f["nearest_restaurant_m"] = near
    return f

print(f"engine ready — {len(POI_TYPES)} POI types, {len(rest)} restaurants")
```

```python
SEED = 42
N_PER_AREA = 300
MIN_DIST = 50.0   # drop points closer than this to a real restaurant

rng = np.random.default_rng(SEED)
bg_points = []
for area, (clat, clon) in CENTERS.items():
    cx, cy = to_xy(clat, clon)
    made, tries = 0, 0
    while made < N_PER_AREA and tries < N_PER_AREA * 20:
        k = (N_PER_AREA - made) * 2
        r  = AREA_RADIUS * np.sqrt(rng.random(k))   # sqrt => uniform in circle
        th = 2 * np.pi * rng.random(k)
        xs, ys = cx + r*np.cos(th), cy + r*np.sin(th)
        d, _ = rest_tree.query(np.c_[xs, ys], k=1)  # dist to nearest real restaurant
        for x, y, dist in zip(xs, ys, d):
            if made >= N_PER_AREA: break
            tries += 1
            if dist < MIN_DIST:  # too close to a real restaurant -> skip
                continue
            lat, lon = to_ll(x, y)
            bg_points.append({"search_area": area, "latitude": lat, "longitude": lon})
            made += 1

bg_df = pd.DataFrame(bg_points)
print(f"generated {len(bg_df)} background points")
print(bg_df.search_area.value_counts())
bg_df.head()
```

```python
# background rows: compute neighbourhood features for each random point
records = []
for _, p in bg_df.iterrows():
    f = compute_features(p.latitude, p.longitude, exclude_self=False)
    f["place_id"]      = f"BG_{len(records):05d}"   # synthetic id, clearly marked
    f["latitude"]      = p.latitude
    f["longitude"]     = p.longitude
    f["search_area"]   = p.search_area
    f["restaurant_name"] = np.nan   # no restaurant here -> left blank
    f["primary_type"]    = np.nan   # left blank (missing, per your plan)
    f["searched_as"]     = np.nan   # fill later per your guidance
    records.append(f)

background = pd.DataFrame(records)
background.to_csv("../data/processed/background_points.csv", index=False)
print("saved background_points.csv —", background.shape)
background.head()
```

```python
# recompute real restaurants with the SAME engine, so both halves match
real_raw = pd.read_csv(REST_PATH).dropna(subset=["latitude", "longitude"])

recs = []
for _, r in real_raw.iterrows():
    f = compute_features(r.latitude, r.longitude, exclude_self=True)  # exclude self!
    f["place_id"]    = r.place_id
    f["latitude"]    = r.latitude
    f["longitude"]   = r.longitude
    f["search_area"] = r.search_area
    f["primary_type"] = r.primary_type
    f["searched_as"]  = r.searched_as
    recs.append(f)

real_features = pd.DataFrame(recs)
real_features.to_csv("../data/processed/real_features.csv", index=False)
print("recomputed real —", real_features.shape)
```

# merging

```python
real = pd.read_csv("../data/processed/real_features.csv")
bg   = pd.read_csv("../data/processed/background_points.csv")
bg = bg.drop(columns=["restaurant_name"], errors="ignore")

# columns common to both, BEFORE adding label
common = [c for c in real.columns if c in bg.columns]

real2 = real[common].copy(); real2["label"] = 1
bg2   = bg[common].copy();   bg2["label"]   = 0

merged = pd.concat([real2, bg2], ignore_index=True)
merged = merged.loc[:, ~merged.columns.duplicated()]   # safety
merged.to_csv("../data/processed/dataset_labeled.csv", index=False)

print("merged —", merged.shape)
print(merged["label"].value_counts())
```
