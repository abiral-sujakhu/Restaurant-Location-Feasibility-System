```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../data/processed/final_dataset.csv")

# Display the first 5 rows
df.head()
```

```python
df.describe()
```

## Check dataset info

```python
df.info()
```

### Check for missing values

```python
missing = df.isnull().sum()
missing = missing[missing > 0]
print(missing)
```

### Dropping the colummn price_level

```python
df.drop(columns=["price_level"], inplace=True)
```

### Dropping the rows with missing values

```python
# Drop rows with missing values in the specified columns
df = df.dropna(subset=[
    "user_rating_count",
    "avg_restaurant_rating_500m",
    "avg_review_ratings_500m",
    "reviews",
]).reset_index(drop=True)

# Verify the result
print("New shape:", df.shape)
print("Remaining missing values:")
print(df[[
    "user_rating_count",
    "avg_restaurant_rating_500m",
    "avg_review_ratings_500m",
    "reviews",
]].isnull().sum())
```

```python
df.info()
```

### printing the name of restaurant with the missing primary type for manual imputation

```python
# Show rows where primary_type is missing
df[df["primary_type"].isna()][["restaurant_name", "primary_type"]]
```

### all the unique  primary type

```python
# Show all unique primary types
print(df["primary_type"].dropna().unique())
```

### Manual Imputation of Missing primary_type Values


```python
df.loc[df["restaurant_name"] == "Cafe Mitini", "primary_type"] = "cafe"

df.loc[df["restaurant_name"] == "Titanic Dance Bar", "primary_type"] = "bar"

df.loc[df["restaurant_name"] == "Arabica Brew Coffee School", "primary_type"] = "coffee_shop"

df.loc[df["restaurant_name"] == "Srisha's Dairy and Bakery", "primary_type"] = "bakery"

df.loc[df["restaurant_name"] == "The Baristas Coffee School (Kumaripati Branch)", "primary_type"] = "coffee_shop"

df.loc[df["restaurant_name"] == "The Baristas Coffee School (Lazimpat Branch)", "primary_type"] = "coffee_shop"
```

### Check for Duplicate Rows

```python
print("Duplicate rows:", df.duplicated().sum())
```

#### Check Data Types

```python
df.info()
```

### Data Validation

```python
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns

for col in numeric_cols:
    print(f"{col}: Min = {df[col].min()}, Max = {df[col].max()}")
```

```python
# Save the cleaned dataset
df.to_csv("../data/processed/cleaned_final_dataset.csv", index=False)

print("Dataset saved successfully as 'cleaned_final_dataset.csv'")
```
