```python
meta_cols = [
    "place_id",
    "latitude",
    "longitude",
    "search_area",
    "primary_type",
    "searched_as",
    "label",
]

score_cols = [
    "bank_count_500m",
    "bus_stop_count_500m",
    "cinema_count_500m",
    "clinic_count_500m",
    "college_count_500m",
    "hospital_count_500m",
    "museum_count_500m",
    "office_count_500m",
    "parking_space_count_500m",
    "recreation_count_500m",
    "retail_count_500m",
    "school_count_500m",
    "temple_count_500m",
    "nearest_restaurant_m",
    "competitor_count_500m",
    "avg_restaurant_rating_500m",
    "avg_review_ratings_500m",
]

result_cols = [
    "feasibility_score",
    "feasibility_label",
    "cluster",
]

ordered_cols = meta_cols + result_cols

for col in score_cols:
    ordered_cols.extend([
        col,
        f"{col}_normalized",
        f"{col}_contribution",
    ])

other_cols = [col for col in df.columns if col not in ordered_cols]

df = df[ordered_cols + other_cols]
```
